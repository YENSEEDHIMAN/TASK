```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np

xpoints = np.array([-6, 6])
ypoints = np.array([50, 250])

plt.plot(xpoints, ypoints)
plt.show()
```

```python
xpoints=np.array([8,9])
ypoints=np.array([8,9])
plt.plot(xpoints,ypoints)
plt.show()
```

```python
xpoints=np.array([1,9])
ypoints=np.array([1,9])
xdot=np.array([-1,7])
ydot=np.array([1,5])
plt.plot(xdot,ydot,'g')
plt.plot(xpoints,ypoints,'y')
plt.show()
```

```python
xpoints=np.array([1,9])
ypoints=np.array([1,9])
xdot=np.array([-1,7])
ydot=np.array([-1,5])
plt.plot(xdot,ydot,'o')
plt.plot(xpoints,ypoints,'y')
plt.show()
```
