```python
import requests                     # To send HTTP requests 
from bs4 import BeautifulSoup       # To parse HTML content
```

```python

# Send an HTTP GET request to the FDA website
req = requests.get("https://www.fda.gov/")

soup = BeautifulSoup(req.content, "html.parser")  
# Print the raw text content of the entire webpage
print(soup.get_text())
```

```python
res=soup.title          # Extract the title tag from the webpage
print(res.get_text())   # Print the text inside the title tag
```

```python
res=soup.header            # Finds the first <header> tag in the HTML
print(res.get_text())       # Print the text content inside the <header> tag
```

```python
res=soup.footer
print(res.get_text())
```

```python
res = soup.find_all("h3", class_="lcds-card__title")  #Find all <h3> tags with the class "lcds-card__title
for item in res:
    print(item.get_text())
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://chaldal.com/fresh-vegetable"
headers = {
    "User-Agent": "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/132.0.0.0 Safari/537.36",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8",
    "Accept-Encoding": "gzip, deflate, br",
    "Accept-Language": "en-US,en;q=0.9",
    "Connection": "keep-alive"
}

session = requests.Session()
response = session.get(url, headers=headers, allow_redirects=True)

# Check if request was successful
if response.status_code == 200:
    soup = BeautifulSoup(response.content, 'html.parser')
    res = soup.find_all(class_="name")
    
    if res:
        for item in res:
            text = item.get_text(separator="\n", strip=True)  # Remove extra spaces & join lines
            print(text)
    else:
        print("No products found. The content might be dynamically loaded.")
else:
    print(f"Failed to retrieve page. Status code: {response.status_code}")
```
